```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
ds = pd.read_csv('Social_Network_Ads.csv')
X = ds.iloc[:, :-1].values
y = ds.iloc[:, -1].values
```

Spliting Dataset into Training and Testing Set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
print(X_train)
```

```python
print(y_train)
```

```python
print(X_test)
```

```python
print(y_test)
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
print(X_train)
```

```python
print(X_test)
```

Training Logistic Regression model on training set

```python
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression()
classifier.fit(X_train, y_train)
```

```python
print(classifier.predict(sc.transform([[50,40000]])))
```

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn import metrics

y_pred_train  = classifier.predict(X_train)
y_pred_test   = classifier.predict(X_test)
print("Train Accuracy: ", metrics.accuracy_score(y_train, y_pred_train))
print("Test Accuracy: ", metrics.accuracy_score(y_test, y_pred_test))
print("Confusion Matrix for Train:\n", metrics.confusion_matrix(y_train, y_pred_train))
print("Confusion Matrix for Test:\n", metrics.confusion_matrix(y_test, y_pred_test))
print("Classification Report for Train:\n", metrics.classification_report(y_train, y_pred_train))
print("Classification Report for Test:\n", metrics.classification_report(y_test, y_pred_test))
print("F1 Score: ", metrics.f1_score(y_test, y_pred_test))
print("Precision Score: ", metrics.precision_score(y_test, y_pred_test))
print("Recall Score: ", metrics.recall_score(y_test, y_pred_test))
print("ROC AUC Score: ", metrics.roc_auc_score(y_test, y_pred_test))
```

Visulising the Training set results

```python
from sklearn.inspection import DecisionBoundaryDisplay

DecisionBoundaryDisplay.from_estimator(
    classifier,
    X_train,
    response_method="predict",
    cmap=plt.cm.coolwarm,
    alpha=0.8,
    xlabel=ds.columns[0],
    ylabel=ds.columns[1],
)

plt.scatter(X_train[:, 0], X_train[:, 1],
            c=y_train,
            cmap=plt.cm.coolwarm,
            s=20,
            edgecolors="k")
plt.title("Logistic Regression (Training Set)")
plt.show()
```

Visualizing Test set result

```python
from sklearn.inspection import DecisionBoundaryDisplay

DecisionBoundaryDisplay.from_estimator(
    classifier,
    X_test,
    response_method="predict",
    cmap=plt.cm.coolwarm,
    alpha=0.8,
    xlabel=ds.columns[0],
    ylabel=ds.columns[1],
)
plt.scatter(X_test[:,0], X_test[:,1],
            c=y_test,
            cmap=plt.cm.coolwarm,
            s=20,
            edgecolors="k")
plt.title("Logistic Regression (Testing Set)")
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```
